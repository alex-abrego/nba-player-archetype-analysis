```python
import pandas as pd
import numpy as np

import os
import time
from time import sleep
from pathlib import Path

from nba_api.stats.endpoints import (
    leaguedashplayerstats,
    leaguedashptstats,
    leaguedashplayershotlocations,
    leaguehustlestatsplayer
)
from nba_api.stats.static import players as nba_players, teams as nba_teams

SEASON = "2025-26"
SEASON_TYPE = "Regular Season"
```

```python
RAW_DATA_DIR = Path("C:\\Users\\aabre\\Documents\\data-science-projects\\nba-player-archetype-clustering\\data\\raw\\2025_26")
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)
RAW_DATA_DIR
```

```python
def get_base_stats(season = SEASON):
    base_stats_df = leaguedashplayerstats.LeagueDashPlayerStats(
        measure_type_detailed_defense="Base",
        per_mode_detailed="Totals",
        season=season,
        season_type_all_star=SEASON_TYPE,
        timeout=60
    ).get_data_frames()[0]
    base_stats_df.to_csv(RAW_DATA_DIR / f"base_stats.csv", index=False)
    return base_stats_df.columns.tolist(), base_stats_df.head()

get_base_stats()
```

```python
def get_advanced_stats(season = SEASON):
    advanced_stats_df = leaguedashplayerstats.LeagueDashPlayerStats(
        measure_type_detailed_defense="Advanced",
        per_mode_detailed="Totals",
        season=season,
        season_type_all_star=SEASON_TYPE,
        timeout=60
    ).get_data_frames()[0]
    advanced_stats_df.to_csv(RAW_DATA_DIR / f"advanced_stats.csv", index=False)
    return advanced_stats_df.columns.tolist(), advanced_stats_df.head()

get_advanced_stats()
```

```python
def get_scoring_splits(season = SEASON):
    scoring_splits_df = leaguedashplayerstats.LeagueDashPlayerStats(
        measure_type_detailed_defense="Scoring",
        per_mode_detailed="Totals",
        season=season,
        season_type_all_star=SEASON_TYPE,
        timeout=60
    ).get_data_frames()[0]
    scoring_splits_df.to_csv(RAW_DATA_DIR / f"scoring_splits.csv", index=False)
    return scoring_splits_df.columns.tolist(), scoring_splits_df.head()

get_scoring_splits()
```

```python
def get_usage_stats(season = SEASON):
    usage_stats_df = leaguedashplayerstats.LeagueDashPlayerStats(
        measure_type_detailed_defense="Usage",
        per_mode_detailed="Totals",
        season=season,
        season_type_all_star=SEASON_TYPE,
        timeout=60
    ).get_data_frames()[0]
    usage_stats_df.to_csv(RAW_DATA_DIR / f"usage_stats.csv", index=False)
    return usage_stats_df.columns.tolist(), usage_stats_df.head()

get_usage_stats()
```

```python
TRACKING_STATS = ["Passing", "Rebounding", "SpeedDistance", "Possessions", "Defense", "Drives", 
                  "CatchShoot", "PullUpShot", "PostTouch", "ElbowTouch", "PaintTouch", "Efficiency"]
def get_tracking_stats(measure_type = TRACKING_STATS, season = SEASON):
    for measure in measure_type:
        print(f"Collecting {measure} stats...")
        try:
            tracking_stats_df = leaguedashptstats.LeagueDashPtStats(
                pt_measure_type=measure,
                player_or_team="Player",
                per_mode_simple="Totals",
                season=season,
                season_type_all_star=SEASON_TYPE,
                timeout=60
            ).get_data_frames()[0]
            tracking_stats_df.to_csv(RAW_DATA_DIR / f"{measure.lower()}_stats.csv", index=False)
            print(f"Saved {measure.lower()}_stats.csv")
            sleep(1.5)
        except Exception as e:
            print(f"Error occurred while collecting {measure} stats: {e}")
    return [f"{measure.lower()}_stats.csv" for measure in measure_type]

get_tracking_stats()
```

```python
def get_shot_location_stats(season = SEASON):
    shot_location_stats_df = leaguedashplayershotlocations.LeagueDashPlayerShotLocations(
        measure_type_simple="Base",
        per_mode_detailed="Totals",
        distance_range="By Zone",
        season=season,
        season_type_all_star=SEASON_TYPE,
        timeout=60
    ).get_data_frames()[0]
    shot_location_stats_df.to_csv(RAW_DATA_DIR / f"shot_location_stats.csv", index=False)
    return shot_location_stats_df.columns.tolist(), shot_location_stats_df.head()

get_shot_location_stats()
```

```python
def get_hustle_stats(season = SEASON):
    hustle_stats_df = leaguehustlestatsplayer.LeagueHustleStatsPlayer(
        season=season,
        season_type_all_star=SEASON_TYPE,
        per_mode_time="Totals",
        timeout=60
    ).get_data_frames()[0]
    hustle_stats_df.to_csv(RAW_DATA_DIR / f"hustle_stats.csv", index=False)
    return hustle_stats_df.columns.tolist(), hustle_stats_df.head()

get_hustle_stats()
```

```python
datasets = {}
for file in RAW_DATA_DIR.glob("*.csv"):
    dataset_name = file.stem
    if dataset_name.startswith("0"):  # Skips our inventory/summary files
            continue
    datasets[dataset_name] = pd.read_csv(file)
```

```python
shot_location_columns = [
    "PLAYER_ID",
    "PLAYER_NAME",
    "TEAM_ID",
    "TEAM_ABBREVIATION",
    "AGE",
    "NICKNAME",
    "RA_FGM",
    "RA_FGA",
    "RA_FG_PCT",
    "PAINT_NON_RA_FGM",
    "PAINT_NON_RA_FGA",
    "PAINT_NON_RA_FG_PCT",
    "MIDRANGE_FGM",
    "MIDRANGE_FGA",
    "MIDRANGE_FG_PCT",
    "LEFT_CORNER_3_FGM",
    "LEFT_CORNER_3_FGA",
    "LEFT_CORNER_3_FG_PCT",
    "RIGHT_CORNER_3_FGM",
    "RIGHT_CORNER_3_FGA",
    "RIGHT_CORNER_3_FG_PCT",
    "ABOVE_BREAK_3_FGM",
    "ABOVE_BREAK_3_FGA",
    "ABOVE_BREAK_3_FG_PCT",
    "BACKCOURT_FGM",
    "BACKCOURT_FGA",
    "BACKCOURT_FG_PCT",
    "CORNER_3_FGM",
    "CORNER_3_FGA",
    "CORNER_3_PCT"
]

shot_location_clean = datasets["shot_location_stats"].copy()
shot_location_clean.columns = shot_location_columns
shot_location_clean = shot_location_clean.iloc[1:]
datasets["shot_location_stats"] = shot_location_clean

shot_location_clean.to_csv(RAW_DATA_DIR / f"shot_location_stats.csv", index=False)
shot_location_clean.head()
```

```python
dataset_inventory = {name: df.shape for name, df in datasets.items()}
dataset_inventory_df = pd.DataFrame.from_dict(dataset_inventory, orient='index', columns=['Rows', 'Columns'])
dataset_inventory_df
```

```python
column_inventory = {name: df.columns.tolist() for name, df in datasets.items()}
column_inventory_df = pd.DataFrame.from_dict(column_inventory, orient='index')
column_inventory_df
```

```python
collection_summary = []
for name, df in datasets.items():
    collection_summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Unique Players": df['PLAYER_ID'].nunique() if 'PLAYER_ID' in df.columns else np.nan,
        "Duplicate Players": df['PLAYER_ID'].duplicated().sum() if 'PLAYER_ID' in df.columns else np.nan
    })
collection_summary = pd.DataFrame(collection_summary)
collection_summary
```

```python
dataset_inventory_df.to_csv(RAW_DATA_DIR / "01_dataset_inventory.csv", index=False)
column_inventory_df.to_csv(RAW_DATA_DIR / "02_column_inventory.csv", index=False)
collection_summary.to_csv(RAW_DATA_DIR / "03_collection_summary.csv", index=False)
```
